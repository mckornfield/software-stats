# India Data Collection

Loads embedded NASSCOM/MOSPI 2023 data from `src/india_data.py` and writes `data/processed/merged_india_data.csv`.

```python
import sys
from pathlib import Path
ROOT = Path().resolve().parent
sys.path.insert(0, str(ROOT / 'src'))

import pandas as pd
from india_data import get_merged_india_data
```

```python
df = get_merged_india_data()
print(f'India data: {len(df)} rows, {len(df.columns)} columns')
print(f'Roles: {sorted(df["role"].unique())}')
print(f'SWE mid salary (INR): {df[(df.role=="software_engineer")&(df.career_stage=="mid")]["median_salary_local"].values[0]:,}')
print(f'SWE mid salary (PPP USD): {df[(df.role=="software_engineer")&(df.career_stage=="mid")]["median_salary_ppp_usd"].values[0]:,.0f}')
df.head()
```

```python
out = ROOT / 'data' / 'processed' / 'merged_india_data.csv'
df.to_csv(out, index=False)
print(f'Wrote {len(df)} rows to {out}')
```
