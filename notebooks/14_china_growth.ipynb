## China Wage and Employment Growth (2019–2023)

NBS median wage and employment by role, indexed to 2019=100.

```python
import sys
from pathlib import Path
ROOT = Path().resolve().parent
sys.path.insert(0, str(ROOT / 'src'))
import pandas as pd
import plotly.express as px

ROLE_LABELS = {
    'software_engineer': 'Software Engineer', 'lawyer': 'Lawyer',
    'physician': 'Physician', 'financial_analyst': 'Financial Analyst',
    'registered_nurse': 'Registered Nurse', 'civil_engineer': 'Civil Engineer',
    'construction_laborer': 'Construction Laborer', 'farm_worker': 'Farm Worker',
    'manufacturing_worker': 'Manufacturing Worker', 'retail_worker': 'Retail Worker',
}
YEARS = [2019, 2020, 2021, 2022, 2023]

df = pd.read_csv(ROOT / 'data' / 'processed' / 'merged_china_data.csv')
grw = df[df['career_stage'] == 'mid'].copy()
grw['role_label'] = grw['role'].map(ROLE_LABELS)
```

```python
rows = []
for _, row in grw.iterrows():
    base = row['wage_2019']
    for yr in YEARS:
        rows.append({'role_label': row['role_label'], 'year': yr,
                     'wage_index': row[f'wage_{yr}'] / base * 100})
wage_long = pd.DataFrame(rows)

fig = px.line(
    wage_long, x='year', y='wage_index', color='role_label', markers=True,
    title='China Wage Growth by Role, 2019–2023 (2019 = 100, CNY)',
    labels={'wage_index': 'Wage Index (2019=100)', 'year': 'Year', 'role_label': 'Role'},
)
fig.add_hline(y=100, line_dash='dot', line_color='grey')
fig.show()
```

```python
rows2 = []
for _, row in grw.iterrows():
    base = row['emp_2019']
    for yr in YEARS:
        rows2.append({'role_label': row['role_label'], 'year': yr,
                      'emp_index': row[f'emp_{yr}'] / base * 100})
emp_long = pd.DataFrame(rows2)

fig2 = px.line(
    emp_long, x='year', y='emp_index', color='role_label', markers=True,
    title='China Employment Growth by Role, 2019–2023 (2019 = 100)',
    labels={'emp_index': 'Employment Index (2019=100)', 'year': 'Year', 'role_label': 'Role'},
)
fig2.add_hline(y=100, line_dash='dot', line_color='grey')
fig2.show()
```
