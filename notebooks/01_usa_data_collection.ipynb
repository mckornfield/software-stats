# USA Data Collection

Loads embedded BLS OES 2023 data from `src/usa_data.py` and writes `data/processed/merged_usa_data.csv`.

```python
import sys
from pathlib import Path
ROOT = Path().resolve().parent
sys.path.insert(0, str(ROOT / 'src'))

import pandas as pd
from usa_data import get_merged_usa_data
```

```python
df = get_merged_usa_data()
print(f'USA data: {len(df)} rows, {len(df.columns)} columns')
print(f'Roles: {sorted(df["role"].unique())}')
print(f'Career stages: {sorted(df["career_stage"].unique())}')
df.head()
```

```python
out = ROOT / 'data' / 'processed' / 'merged_usa_data.csv'
df.to_csv(out, index=False)
print(f'Wrote {len(df)} rows to {out}')
print(df.columns.tolist())
```
